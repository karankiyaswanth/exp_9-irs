```python
import spacy
import re
from collections import Counter

nlp = spacy.load("en_core_web_sm")

def clean_tweet(text):
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"#\w+", "", text)
    text = re.sub(r"[^A-Za-z0-9\s]", "", text)
    text = re.sub(r" ", " ", text)
    return text.strip()

def mine_tweets(keyword, since, until, max_tweets=200):
    tweets = [
        "Climate change is creating a global crisis.",
        "We need climate action to protect the future.",
        "Global warming is increasing carbon emissions.",
        "Climate change affects the future of Earth.",
        "NASA reports that global warming is affecting Earth.",
        "Greta Thunberg called for climate action.",
        "The UN discussed climate change and global emissions.",
        "Biden announced new climate action.",
        "India is working to reduce carbon emissions.",
        "Amazon announced plans to reduce emissions.",
        "Shell faces pressure to reduce carbon emissions.",
        "New York is preparing for climate change.",
        "COP28 discussed global warming and climate action.",
        "Climate crisis requires global action.",
        "The future of Earth depends on climate action."
    ]
    tweets = (tweets * 14)[:max_tweets]
    return tweets

def analyze_tweets(tweets):
    cleaned_tweets = [clean_tweet(tweet) for tweet in tweets]
    all_text = " ".join(cleaned_tweets)
    words = all_text.lower().split()
    stop_words = {
        "the", "is", "a", "an", "and", "to", "of", "in", "for", "on", "with",
        "this", "that", "are", "we", "our", "it", "be", "one", "now", "from",
        "as", "by", "has", "have", "its", "they", "their", "at", "or", "will", "can"
    }
    filtered_words = [
        word for word in words if len(word) > 3 and word not in stop_words
    ]
    word_freq = Counter(filtered_words).most_common(10)

    entity_counter = Counter()
    for text in tweets:
        doc = nlp(text)
        for ent in doc.ents:
            if ent.label_ in ["PERSON", "ORG", "GPE", "LOC"]:
                entity_counter[ent.text] += 1
    return word_freq, entity_counter.most_common(10)

print("Twitter Mining and Trend Analysis")
print("----------------------------------")
keyword = input("Enter keyword or hashtag: ").strip()
since = input("Enter start date (YYYY-MM-DD): ").strip()
until = input("Enter end date (YYYY-MM-DD): ").strip()

tweets = mine_tweets(keyword, since, until, max_tweets=200)
print(f"\nCollected {len(tweets)} tweets.")

trends, entities = analyze_tweets(tweets)

print("\n--- Trending Words ---")
for word, frequency in trends:
    print(f"{word}: {frequency}")

print("\n--- Named Entities (People, Orgs, Places) ---")
for entity, frequency in entities:
    print(f"{entity}: {frequency}")
```
